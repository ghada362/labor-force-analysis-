```python
import pandas as pd
import numpy as np
```

```python
INPUT_FILE = "member_info_2024_last.csv"
```

```python
converted = 0
skipped = 0

for col, mapping in MAPPINGS.items():

    if col in df.columns:

        df[col] = df[col].map(mapping).fillna(df[col])

        print(f"✅ {col}")

        converted += 1

    else:

        skipped += 1

print(f"\n✅ تم تحويل {converted} عمود")
print(f"⚠️ غير موجود {skipped} عمود")
```

```python
empty_after = df.columns[df.isna().all()]

if len(empty_after) > 0:

    print("\n🗑️ حذف أعمدة فارغة بعد التنظيف:")

    for col in empty_after:
        print(f" - {col}")

    df.drop(columns=empty_after, inplace=True)
```

```python
def classify_occupation(text):
    if pd.isna(text):
        return "غير محدد"
 
    text = str(text).strip()
 
    rules = [
        # --- زراعة وثروة حيوانية ---
        ("زراعة وثروة حيوانية", [
            "زراعي", "مزارع", "فلاح", "حصاد", "راعي",
            "صياد", "ري", "مشتل", "دواجن", "مواشي",
            "زراعه", "غاب", "حقل",
        ]),
 
        # --- بناء وتشييد ---
        ("بناء وتشييد", [
            "بناء", "معماري", "نجار", "حداد", "سباك",
            "دهان", "بلاط", "مقاول", "تشطيب", "عازل",
            "حفار", "مبيض", "محارة", "مسلح",
        ]),
 
        # --- نقل ومواصلات ---
        ("نقل ومواصلات", [
            "سائق", "توك توك", "توكتوك", "ميكروباص",
            "شاحنة", "شاحنه", "تاكسي", "نقل", "طيار",
            "بحار", "اتوبيس",
        ]),
 
        # --- تعليم ---
        ("تعليم", [
            "مدرس", "معلم", "أستاذ", "استاذ", "محاضر",
            "مدرب تعليم", "ناظر", "مشرف تربوي",
        ]),
 
        # --- صحة ---
        ("صحة وطب", [
            "طبيب", "ممرض", "ممرضة", "صيدل", "أسنان",
            "مختبر", "أشعة", "تمريض", "فيزياء طبية",
            "صحة", "مستشفى",
        ]),
 
        # --- هندسة ---
        ("هندسة", [
            "مهندس", "هندسة",
        ]),
 
        # --- إدارة وأعمال مكتبية ---
        ("إدارة وأعمال مكتبية", [
            "موظف", "مدير", "سكرتير", "محاسب",
            "إداري", "اداري", "بنك", "تأمين",
            "موارد بشرية", "مراجع",
        ]),
 
        # --- تجارة ومبيعات ---
        ("تجارة ومبيعات", [
            "بائع", "تاجر", "مندوب مبيعات", "مندوب",
            "كاشير", "موزع", "مسوق",
        ]),
 
        # --- صناعة وإنتاج ---
        ("صناعة وإنتاج", [
            "إنتاج", "انتاج", "فني", "ميكانيكي",
            "كهربائي", "مشغل", "صانع", "تقني",
            "مصنع", "ورشة",
        ]),
 
        # --- أمن وشرطة وجيش ---
        ("أمن وشرطة وجيش", [
            "أمن", "امن", "شرطة", "شرطه", "ضابط",
            "جندي", "عسكري", "حارس", "مجند",
        ]),
 
        # --- خدمات وسياحة ---
        ("خدمات وسياحة", [
            "خدمات", "نظافة", "نظافه", "نادل", "طباخ",
            "شيف", "فندق", "سياحه", "سياحة", "حلاق",
            "غسيل",
        ]),
 
        # --- قانون وإعلام ---
        ("قانون وإعلام وفنون", [
            "محامي", "قاضي", "نيابة", "صحفي", "إعلام",
            "اعلام", "فنان", "ممثل", "مصور",
        ]),
 
        # --- تقنية معلومات ---
        ("تقنية معلومات", [
            "كمبيوتر", "برمجة", "برمجه", "مبرمج",
            "شبكات", "it", "تقنية", "سوفت",
        ]),
 
        # --- أخصائي اجتماعي ونفسي ---
        ("خدمة اجتماعية ونفسية", [
            "أخصائي", "اخصائي", "اجتماعي", "نفسي",
            "نفس", "إرشاد",
        ]),
    ]
 
    text_lower = text.lower()
    for group, keywords in rules:
        for kw in keywords:
            if kw in text_lower:
                return group
 
    return "أخرى"
 
df["OCCUPATION_GROUP"] = df["CURRENT_WORK_CLEAN"].apply(classify_occupation)
```

```python
import pandas as pd
import re

# ============================================================
# توحيد الإملاء بالـ Regex
# ============================================================
def normalize(text):
    if pd.isna(text): return text
    t = str(text).strip()
    t = re.sub(r'\s+', ' ', t)

    # توحيد حروف
    t = t.replace('ة', 'ه').replace('أ','ا').replace('إ','ا').replace('آ','ا')
    t = t.replace('ى', 'ي')
    return t.strip()

df["JOB_NORM"] = df["CURRENT_WORK"].apply(normalize)

# ============================================================
# دالة التصنيف بالـ Keywords
# ============================================================
GROUPS = [
    ("زراعة وثروة حيوانية", [
        r"زراع", r"مزارع", r"فلاح", r"محصول", r"حصاد", r"حرث",
        r"راعي", r"مواشي", r"دواجن", r"فراخ", r"طيور", r"اغنام",
        r"ماشيه", r"صياد", r"صيد", r"سمك", r"بسته", r"قطن",
        r"ذره", r"برسيم", r"قمح", r"بصل", r"غلال", r"حبوب",
        r"مشتل", r"جنايني", r"جناين", r"نخيل", r"ري الارض",
    ]),
    ("بناء وتشييد", [
        r"بنا[ءا]", r"معمار", r"مبلط", r"نقاش", r"محاره", r"محارة",
        r"سبا[كق]", r"حداد", r"نجار مسلح", r"مسلح", r"خرسان",
        r"رخام", r"سيرامي[كق]", r"بلاط", r"جبس", r"تشطيب",
        r"تشيد", r"تشييد", r"دهان", r"نقاش", r"طوب",
        r"لحام", r"كهربائي مناز", r"كهربا[ءء] مناز",
    ]),
    ("نقل ومواصلات", [
        r"سائق", r"سواق", r"توك\s*توك", r"توكتوك", r"ميكروباص",
        r"مكروباص", r"تاكسي", r"نقل ركاب", r"نقل بضائع",
        r"تروسيكل", r"تريسكل", r"دليفر", r"دلفر", r"توصيل",
        r"شاحنه", r"لودر", r"حفار", r"جرار", r"قطار",
    ]),
    ("تعليم", [
        r"مدرس", r"معلم", r"مدرسه", r"معلمه", r"استاذ",
        r"محفظ", r"تحفيظ", r"حضانه", r"روضه", r"تاسيس",
        r"موجه", r"مشرف.*تعليم", r"دكتور.*جامعه",
        r"معيد", r"محاضر",
    ]),
    ("صحة وطب", [
        r"طبيب", r"طبيبه", r"دكتور", r"دكتوره", r"ممرض",
        r"ممرضه", r"صيدل", r"علاج طبيعي", r"اشعه",
        r"تحاليل طبي", r"مسعف", r"تمريض", r"عناية مركزه",
        r"اسنان", r"جراح",
    ]),
    ("هندسة", [
        r"مهندس", r"مهندسه",
    ]),
    ("إدارة وأعمال مكتبية", [
        r"محاسب", r"محاسبه", r"موظف", r"موظفه", r"مدير",
        r"مديره", r"سكرتير", r"سكرتيره", r"كاتب", r"كاتبه",
        r"اداري", r"اداريه", r"مراجع حسابات", r"كاشير",
        r"تلر", r"بنك", r"شؤون عاملين", r"موارد بشريه",
        r"مشرف اداري", r"مسئول",
    ]),
    ("تجارة ومبيعات", [
        r"بائع", r"بائعه", r"بياع", r"بياعه", r"تاجر",
        r"تاجره", r"مندوب مبيعات", r"مبيعات", r"سيلز",
        r"موزع", r"بقال", r"بقاله",
    ]),
    ("صناعة وإنتاج", [
        r"عامل انتاج", r"عامل إنتاج", r"فني", r"فنيه",
        r"ميكانيك", r"كهربائي.*صناع", r"مصنع",
        r"خراط", r"براد", r"لحام.*معادن", r"خياط",
        r"ترزي", r"خياطه", r"صنايعي", r"صنايعيه",
    ]),
    ("أمن وشرطة وجيش", [
        r"امن", r"أمن", r"شرطه", r"شرطة", r"ضابط",
        r"جندي", r"عسكري", r"حارس", r"غفير", r"مجند",
        r"حراسه",
    ]),
    ("خدمات وسياحة وفندقة", [
        r"نظافه", r"نظافة", r"طباخ", r"طباخه", r"شيف",
        r"فندق", r"سياح", r"حلاق", r"كوافير", r"نادل",
        r"مطعم", r"قهوجي", r"قهوه", r"مشروبات",
    ]),
    ("قانون وإعلام وفنون", [
        r"محامي", r"محاميه", r"قاضي", r"صحفي", r"صحفيه",
        r"اعلام", r"فنان", r"ممثل", r"مصور", r"فوتوغراف",
    ]),
    ("تقنية معلومات", [
        r"مبرمج", r"برمجه", r"كمبيوتر.*مهندس",
        r"مهندس.*كمبيوتر", r"شبكات", r"تقنيه",
        r"it\b", r"ايتي",
    ]),
    ("خدمة اجتماعية ونفسية", [
        r"اخصائي", r"اخصائيه", r"أخصائي", r"أخصائيه",
        r"اجتماعي", r"اجتماعيه", r"نفسي", r"نفسيه",
        r"ارشاد",
    ]),
    ("دين وأوقاف", [
        r"امام", r"إمام", r"مؤذن", r"مأذون", r"خطيب",
        r"اوقاف", r"مسجد", r"كاهن", r"قسيس",
    ]),
]

def classify(text):
    if pd.isna(text): return "غير محدد"
    t = str(text)
    for group, patterns in GROUPS:
        for pat in patterns:
            if re.search(pat, t, re.IGNORECASE):
                return group
    return "أخرى"

print("🔍 تصنيف الوظائف...")
df["OCCUPATION_GROUP"] = df["JOB_NORM"].apply(classify)

# fallback: لو أخرى ياخد قيمة JOB_NORM
df["OCCUPATION_GROUP"] = df.apply(
    lambda row: row["JOB_NORM"]
    if row["OCCUPATION_GROUP"] == "أخرى"
    else row["OCCUPATION_GROUP"],
    axis=1
)

# ============================================================
# إحصائيات
# ============================================================
print("\n📊 توزيع المجموعات المهنية (للمشتغلين فقط):")
employed = df[df["WORK_LAST_WEEK_CODE"] == "نعم"]
counts = employed["OCCUPATION_GROUP"].value_counts()
total = len(employed)
print(f"{'المجموعة':<35} {'عدد':>8}  {'نسبة':>7}")
print("-"*55)
for grp, cnt in counts.head(20).items():
    label = grp[:33] if len(grp) > 33 else grp
    print(f"{label:<35} {cnt:>8,}  {cnt/total*100:>6.1f}%")

unclassified = (employed["OCCUPATION_GROUP"] == "غير محدد").sum()
print(f"\n{'غير محدد':<35} {unclassified:>8,}  {unclassified/total*100:>6.1f}%")
print(f"\n✅ إجمالي المشتغلين: {total:,}")

```
